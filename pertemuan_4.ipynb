```python
 #1
tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)
pesan =   {True:"merupakan tahun kabisat", False:"bukan tahun kabisat"}
print(tahun, pesan[kabisat])
```

```python
#2
judul = "PROGRAM HITUNG PERSEGI PANJANG"

# Fungsi untuk menghitung luas dan keliling
def hitung_persegi_panjang():
    panjang = float(input("Masukkan panjang: "))
    lebar = float(input("Masukkan lebar: "))

    luas = panjang * lebar
    keliling = 2 * (panjang + lebar)

    print("Luas persegi panjang =", luas)
    print("Keliling persegi panjang =", keliling)

# Program utama
print(judul)
hitung_persegi_panjang()
```
