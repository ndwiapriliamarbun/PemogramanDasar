```python
#Nama : Nabila Dwi Aprilia Marbun
#NIM : 4232501027
#Mata kuliah : Pemograman Dasar
#Soal 9. Pembacaan Vibrasi dengan Sensor Rusak


data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for n, teks in enumerate(data, start=1):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")

    if nilai > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")

#data = [...] adalah list berisi 5 pembacaan vibrasi dalam bentuk teks. Sensor ke-3 rusak, jadi isinya "ERR".
#jumlah_tidak_valid = 0 adalah penghitung data rusak. Mulainya dari 0.
#for n, teks in enumerate(data, start=1) mengulang untuk tiap data. enumerate memberi nomor unit mulai dari 1 (n) sekaligus isi datanya (teks). Tanpa start=1, nomornya mulai dari 0.
#try: nilai = float(teks) mencoba mengubah teks menjadi angka.
#except ValueError: dijalankan jika langkah 4 gagal (misalnya "ERR"). Di situ program mencetak pesan, menambah penghitung, lalu continue untuk lompat ke unit berikutnya.
#if nilai > 2.0 mencetak Warning jika getaran tinggi.
#if nilai > 3.0 mencetak KRITIS lalu break menghentikan seluruh loop.
#print(f"Jumlah ...") ada di luar loop, jadi tampil sekali di akhir.
```

```python
# CHALLENGE - Pemantauan Beban Harian Pembangkit
kapasitas = 100.0   # MW
beban_jam = ["62.5", "78.0", "91.4", "ERR", "104.2", "85.0"]

beban_valid = []
jumlah_overload = 0
jumlah_error = 0

for jam, teks in enumerate(beban_jam, start=1):
    try:
        beban = float(teks)
    except ValueError:
        print(f"Jam {jam}: data sensor tidak valid")
        jumlah_error += 1
        continue

    beban_valid.append(beban)
    persen = (beban / kapasitas) * 100

    if persen > 100:
        print(f"Jam {jam}: {beban} MW ({persen:.1f}%) -> OVERLOAD")
        jumlah_overload += 1
    elif persen >= 85:
        print(f"Jam {jam}: {beban} MW ({persen:.1f}%) -> WARNING beban tinggi")
    else:
        print(f"Jam {jam}: {beban} MW ({persen:.1f}%) -> NORMAL")

rata_rata = sum(beban_valid) / len(beban_valid)
print(f"Rata-rata beban = {rata_rata:.2f} MW dari {len(beban_valid)} data valid")
print(f"Overload: {jumlah_overload} kali | Data error: {jumlah_error}")


#kapasitas menyimpan kapasitas pembangkit (100 MW) sebagai patokan. beban_jam adalah list berisi 6 pembacaan per jam, berbentuk teks.
#Tiga variabel awal menyiapkan penampung. beban_valid adalah list kosong untuk data yang berhasil dibaca, sedangkan jumlah_overload dan jumlah_error adalah penghitung.
#for ... enumerate(..., start=1) mengulang tiap pembacaan dengan nomor jam mulai dari 1.
#try-except ValueError mencoba mengubah teks ke angka. Jika gagal ("ERR"), program mencetak pesan, menambah jumlah_error, lalu continue ke jam berikutnya.
#beban_valid.append(beban) menyimpan data yang berhasil dibaca ke list.
#persen = (beban / kapasitas) * 100 menghitung beban sebagai persen dari kapasitas.
#if-elif-else mengklasifikasikan: di atas 100% berarti OVERLOAD, 85% ke atas berarti WARNING, selain itu NORMAL. Urutannya penting, yang paling parah dicek lebih dulu.
#Di akhir, sum(beban_valid) / len(beban_valid) menghitung rata-rata dari data valid saja, lalu ringkasan dicetak.
```
