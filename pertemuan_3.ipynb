#1. Urutan Prioritas Operator (Operator Precedence)
 Di dalam Python, operator memiliki tingkatan prioritas yang berbeda saat dieksekusi. Operator perkalian (*) memiliki prioritas yang lebih tinggi daripada operator penjumlahan (+). Oleh karena itu, pada ekspresi 10 + 5 * 2, Python akan mengeksekusi operasi 5 * 2 terlebih dahulu (menghasilkan 10), baru kemudian menambahkannya dengan 10 awal (10 + 10 = 20). Jika kita ingin penjumlahan dilakukan terlebih dahulu, kita wajib menggunakan tanda kurung: (10 + 5) * 2.

 sumber : Python Official Documentation (Section 6.17 - Operator precedence) atau w3schools Python Operators.

#2. Operator Relasional vs Operator Logika
 *   Operator relasional (>, <, >=, <=, ==, !=) membandingkan dua nilai dan menghasilkan satu nilai boolean (True/False) — misalnya suhu > 100.

*  Operator logika (and, or, not) menggabungkan satu atau lebih hasil boolean (biasanya hasil dari operator relasional) menjadi satu keputusan akhir.

contoh : mahasiswa harus memiliki nilai tugas ≥ 60 dan kehadiran ≥ 75%. Kedua operator relasional digunakan untuk membandingkan nilai, sedangkan and digunakan untuk memastikan kedua persyaratan terpenuhi.

nilai = 75

kehadiran = 80

boleh_ujian = (nilai >= 60) and (kehadiran >= 75)

#3.Contoh kasus nyata bug akibat salah paham operator precedence

Kasus ini terdokumentasi sebagai CWE-783 (Operator Precedence Logic Error) dalam basis data kelemahan perangkat lunak yang dikelola MITRE. Kasusnya terjadi pada fungsi autentikasi pengguna, di mana kode dimaksudkan mengembalikan status "gagal login" secara default kecuali password terbukti benar. Karena urutan evaluasi operator dalam ekspresi tidak sesuai dengan maksud programmer, variabel status login justru selalu ditimpa menjadi "sukses" tanpa memandang hasil pengecekan password. Akibatnya, siapa pun bisa dianggap berhasil login meskipun memasukkan password yang salah — celah keamanan yang serius. Perbaikannya adalah menambahkan tanda kurung eksplisit agar pengecekan kondisi gagal benar-benar dievaluasi dan disimpan ke variabel status sebelum nilai "sukses" ditetapkan, bukan mengandalkan urutan prioritas default operator.

```python
# Nama  : Nabila Dwi Aprilia Marbun
# NIM   : 4232501027
# Tanggal: 26 September 2026
# Tujuan: Menghitung operasi aritmatika dari dua bilangan bulat

bilangan1 = int(input("Masukkan bilangan pertama: "))
bilangan2 = int(input("Masukkan bilangan kedua: "))

print("Hasil penjumlahan (+):", bilangan1 + bilangan2)
print("Hasil pengurangan (-):", bilangan1 - bilangan2)
print("Hasil perkalian (*):", bilangan1 * bilangan2)
print("Hasil pembagian (/):", bilangan1 / bilangan2)
print("Hasil pembagian bulat (//):", bilangan1 // bilangan2)
print("Hasil modulus (%):", bilangan1 % bilangan2)
```
